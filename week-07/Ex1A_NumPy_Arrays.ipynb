## Looping through Arrays

```python
arrA = np.array([[
    [11,12,13],
    [14,15,16]
    ]])
for x in arrA:
    print(x * 2)

for x in arrA:
    for y in x:
        print(y * 2)
```

Printing x in arrA seems to multiply the array by 2, so the entire array is printed
Looping through the two dimensions multiplies each number individually

```python
for x in arrA:
    for y in x:
        loop_result = x * 2
        print(loop_result)
```

This will print the array twice

 ## Joining and Stacking Arrays
 
 

```python
joined_array_1 = np.concatenate((arrA, arrA), axis = 0)
print(joined_array_1)
joined_array_2 = np.concatenate((arrA, arrA), axis = 1)
print(joined_array_2)
print(" ____________________________________________")
stacked_array_1 = np.stack((joined_array_1,joined_array_1))
print(stacked_array_1)
print(" ____________________________________________")
#stacked_array_2 = np.stack((joined_array_1, joined_array_2)) ('all input arrays must have the same shape')
#print(stacked_array_2)
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA))
print(another_joined_array)
#another_2 = np.concatenate((joined_array_1, arrA), axis = 1)
#ValueError: all the input array dimensions except for the concatenation axis must match exactly, but along dimension 0, the array at index 0 has size 2 and the array at index 1 has size 1
```

```python
horizontal = np.hstack((arrA, arrA))
print(horizontal)
print("________________________________________")
vertical = np.vstack((arrA, arrA))
print(vertical)
print("________________________________________")
depth = np.dstack((arrA, arrA))
print(depth)
```

## Splitting Arrays

```python
splitsky = np.array_split(joined_array_1, 2)

print(splitsky)
#2 arrays and 2 dimensions
```

```python
spliterino = np.array_split(joined_array_1, 3)

print(spliterino)
# 2 arrays and 3 dimensions
```

## Searching and Replacing Array Values

```python
result_1 = np.where(arrA % 2 == 0, arrA, 'odd')

print(result_1)
```

## Sorting and Filtering Arrays

```python
arrB = np.array([[7, 6],
                 [13, 5],
                 [7, 3]])

arrB_sorted = np.sort(arrB)

print(arrB_sorted)
```

```python
print(arrB)

z = arrB < 7

arrC = arrB[z]

print(arrC)
```

```python
print(arrB)

y = arrB >= 7

arrD = arrB[y]

print(arrD)
```

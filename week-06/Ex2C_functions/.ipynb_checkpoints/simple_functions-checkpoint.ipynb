## LAB 1
Working with defining custom functions

```python
def favorite_things(name, movie, music):
    return f"My name is {name}.\nMy favorite movie is {movie}.\nMy favorite musician/band is {music}."
```

```python
print(favorite_things("Henry", "Generic Movie", "Generic Music"))
```

```python
def why_im_here(purpose):
    return purpose
```

```python
print(why_im_here("I joined the Year Up United Data Analyst program to develop my analytical skills and prepare for a career in data analytics."))
```

```python
def favorite_place(place, why):
    return f"One of my favorite places is {place} because {why}."
```

```python
print(favorite_place("Japan", "I enjoy the culture, food, and beautiful scenery"))
```

```python
print(favorite_place("New York", "I enjoy the city atmosphere and the variety of things to do"))
```

### Array Basics


```python
fruits = ['apple', 'banana', 'cherry', 'date']
arr_1 = np.array(fruits)
print(f'{arr_1}')
print(f'{type(arr_1)}')
#class numpy.ndarray
```

```python
arr_2 = np.array([
    [10, 15, 12],
    [13, 14, 15]
])

print(arr_2)
print(arr_2.shape)
#[[10 15 12]
#[13 14 15]]
#(2, 3)
```

```python
arr_3 = np.array([
    [1, 3, 5],
    [2, 4, 6]
])
print(arr_3)
arr_4 = np.array([2, 6])
print(arr_3[0,2])
print(arr_4)
```

```python
arr_5 = np.array([
    [1, 1, 1],
    [1, 1, 1],
    [1, 1, 1]
])
print(arr_5)
```
